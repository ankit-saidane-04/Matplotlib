```python
# Task 1.

hours = [1, 2, 3, 4, 2, 5, 3, 6, 1, 4]
posts = [5, 8, 12, 15, 9, 20, 13, 25, 4, 17]

plt.scatter(hours, posts, color="purple", s=80)

plt.title("Instagram Usage vs Monthly Posts")
plt.xlabel("Hours Spent on Instagram per Day")
plt.ylabel("Number of Posts per Month")

plt.grid(True)
plt.show()
```

```python
# Task 2.

names = ["Amit", "Rahul", "Priya", "Neha","Jay", "Riya", "Karan", "Pooja",
          "Raj", "Meera", "Dev", "Ankit"]

orders = [3, 5, 8, 4, 10, 6, 7, 2, 9, 5, 11, 6]
spent = [600, 1000, 1800, 800, 2200, 1200,
         1600, 400, 2000, 900, 2500, 1300]

plt.scatter(orders, spent, color="orange", s=80)

for i in range(len(names)):
    plt.annotate(names[i],
        (orders[i], spent[i]),
        xytext=(5, 5),
        textcoords="offset points")

plt.title("Zomato Orders vs Amount Spent")
plt.xlabel("Number of Orders")
plt.ylabel("Total Amount Spent (₹)")

plt.grid(True)
plt.show()
```

```python
# Task 3.

import matplotlib.pyplot as plt
import numpy as np

orders = [3, 5, 8, 4, 10, 6, 7, 2, 9, 5, 11, 6]
spent = [600, 1000, 1800, 800, 2200, 1200,
         1600, 400, 2000, 900, 2500, 1300]

slope, intercept = np.polyfit(orders, spent, 1)

predicted_spent = np.polyval([slope, intercept], orders)

plt.scatter(orders, spent, color="orange", label="Actual data")
plt.plot(orders, predicted_spent, color="red",
         label="Regression line")

plt.title("Zomato Orders vs Spending with Trend Line")
plt.xlabel("Number of Orders")
plt.ylabel("Total Amount Spent (₹)")

plt.legend()
plt.grid(True)
plt.show()

print("Slope:", slope)
print("Intercept:", intercept)
```

```python
# Task 4.
import numpy as np

youtube = [2, 4, 3, 5, 6, 2, 7, 4, 8, 5, 3, 6, 9, 7, 4]
spotify = [3, 5, 4, 6, 7, 3, 8, 5, 9, 6, 4, 7, 10, 8, 5]

slope, intercept = np.polyfit(youtube, spotify, 1)

x_line = np.linspace(min(youtube), max(youtube), 100)
y_line = slope * x_line + intercept

plt.scatter(youtube, spotify, color="green",
            label="Users")
plt.plot(x_line, y_line, color="red",
         label="Regression line")

equation = f"y = {slope:.2f}x + {intercept:.2f}"

plt.text(0.05, 0.95, equation,
              transform=plt.gca().transAxes,
              verticalalignment="top")

plt.title("YouTube vs Spotify Listening Hours")
plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")
plt.legend()
plt.grid(True)
plt.show()

print("Regression equation:", equation)
```

```python
# Task 5.
import numpy as np

orders = [1, 3, 2, 5, 4, 6, 2, 7, 5, 8]
spent = [800, 2500, 1500, 4000, 3200,
         4800, 1800, 5500, 4200, 6500]

# Fit a regression line
slope, intercept = np.polyfit(orders, spent, 1)

# Generate points for a smooth line
x_line = np.linspace(min(orders), max(orders), 100)
y_line = slope * x_line + intercept

# Plot
plt.scatter(orders, spent, color="blue",
            label="Users")
plt.plot(x_line, y_line, color="red",
         label="Regression line")

plt.title("Flipkart Orders vs Total Amount Spent")
plt.xlabel("Number of Flipkart Orders")
plt.ylabel("Total Amount Spent (₹)")
plt.legend()
plt.grid(True)
plt.show()

print(f"Regression equation: y = {slope:.2f}x + {intercept:.2f}")
```
