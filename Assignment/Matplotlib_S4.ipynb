```python
import matplotlib.pyplot as plt
```

```python
# Task 1

apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]
hours = [3, 4, 2, 1, 2]

plt.pie(hours, labels=apps, autopct="%1.1f%%")

plt.title("Daily App Usage")
plt.show()
```

```python
# Task 2.

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [2000, 5000, 1500, 2500]

explode = [0, 0.1, 0, 0]

plt.pie(spending,labels=categories,autopct="%1.1f%%",explode=explode)

plt.title("Monthly Online Spending")
plt.show()
```

```python
# Task 3.
categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [2000, 5000, 1500, 2500]

fig, ax = plt.subplots()

ax.pie(spending,labels=categories,autopct="%1.1f%%")

circle = plt.Circle((0, 0), 0.60, color="white")
ax.add_artist(circle)

ax.set_title("Monthly Online Spending Donut Chart")
plt.show()
```

```python
# Task 4.

teams = ["CSK", "MI", "RCB", "KKR"]
followers = [15, 16, 18, 8]

fig, ax = plt.subplots()

ax.pie(followers,labels=teams,autopct="%1.1f%%",startangle=90)

circle = plt.Circle((0, 0), 0.60, color="white")
ax.add_artist(circle)

ax.set_title("IPL Teams Instagram Followers Share")
ax.axis("equal")

plt.show()
```
