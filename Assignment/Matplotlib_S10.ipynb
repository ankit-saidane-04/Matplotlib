```python
# Task 1.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

users = [120, 135, 128, 150, 165, 180,
         175, 190, 210, 205, 230, 250]

plt.figure(figsize=(10, 5))

plt.plot(months, users, marker="o", color="green")

max_index = users.index(max(users))

plt.annotate(f"Highest: {users[max_index]}K users",
    xy=(months[max_index], users[max_index]),
    xytext=(-80, -35),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", color="red"))

plt.title("Spotify Monthly Active Users")
plt.xlabel("Month")
plt.ylabel("Active Users (Thousands)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Task 2.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

orders = [450, 380, 520, 490, 600, 550,
          430, 650, 580, 700, 620, 750]

plt.figure(figsize=(10, 5))

plt.bar(months, orders, color="orange")

min_index = orders.index(min(orders))

plt.annotate(f"Lowest: {orders[min_index]} orders",
    xy=(min_index, orders[min_index]),
    xytext=(min_index + 1, orders[min_index] + 100),
    arrowprops=dict(arrowstyle="->", color="red"),
    color="red")

plt.title("Zomato Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

```python
# Task 3.
import matplotlib.pyplot as plt

payment_methods = ["UPI", "Card", "Wallet", "Cash"]
percentages = [50, 25, 15, 10]

plt.figure(figsize=(12, 10))

plt.pie(percentages,
    labels=payment_methods,
    autopct="%1.1f%%",
    startangle=90)

plt.legend(payment_methods,
    loc="lower right",
    title="Payment Methods")

plt.title("Flipkart Payment Methods")
plt.show()
```

```python
# Task 4.
import matplotlib.pyplot as plt

movies = ["Movie A", "Movie B", "Movie C",
    "Movie D", "Movie E", "Movie F"]

ratings = [7.5, 8.2, 9.1, 6.8, 8.7, 7.9]
reviews = [1200, 2500, 1800, 900, 3000, 1500]

plt.figure(figsize=(9, 5))

plt.scatter(reviews, ratings, color="blue", s=100)


max_index = ratings.index(max(ratings))

plt.annotate(f"{movies[max_index]}\nRating: {ratings[max_index]}",
    xy=(reviews[max_index], ratings[max_index]),
    xytext=(30, -30),
    textcoords="offset points",
    bbox=dict(boxstyle="round,pad=0.5", fc="yellow", alpha=0.7),
    arrowprops=dict(arrowstyle="->", color="red"))

plt.title("Movie Ratings vs. Number of Reviews")
plt.xlabel("Number of Reviews")
plt.ylabel("Movie Rating")
plt.grid(True)
plt.tight_layout()
plt.show()
```
