```python
# Task 2.
import matplotlib.pyplot as plt

scores = [125, 145, 160, 178, 190, 205, 155, 170, 185, 198,
          132, 148, 165, 180, 195, 210, 140, 158, 175, 188,
          202, 215, 128, 152, 168, 182, 197, 220, 135, 162,
          177, 192, 207, 143, 157, 172, 186, 200, 212, 130,
          150, 166, 184, 196, 218, 138, 154, 173, 189, 204,
          142, 164, 179, 193, 209]

plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.hist(scores, bins=5, color="skyblue", edgecolor="black")
plt.title("IPL Scores - 5 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")

plt.subplot(1, 3, 2)
plt.hist(scores, bins=10, color="lightgreen", edgecolor="black")
plt.title("IPL Scores - 10 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")

plt.subplot(1, 3, 3)
plt.hist(scores, bins=15, color="salmon", edgecolor="black")
plt.title("IPL Scores - 15 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# Task 3.

steps = [5200, 6500, 7200, 8100, 9000, 4500, 6000,
    7500, 8200, 9500, 10500, 6800, 7300, 8800,
    9200, 5000, 6400, 7800, 8500, 11000, 12000,
    5600, 7000, 8000, 9700, 10200, 6200, 7600,
    8900, 11500]

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.hist(steps, bins=6, color="blue", edgecolor="black")
plt.title("Step Counts - Frequency")
plt.xlabel("Daily Steps")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(steps, bins=6, density=True,
         color="purple", edgecolor="black")
plt.title("Step Counts - Density")
plt.xlabel("Daily Steps")
plt.ylabel("Probability Density")

plt.tight_layout()
plt.show()
```
