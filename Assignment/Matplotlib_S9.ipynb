```python
# Task 1.
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato = [120, 150, 130, 170, 200, 250, 230]
swiggy = [100, 140, 120, 160, 180, 220, 210]

plt.bar(days, zomato, color="tomato", label="Zomato")
plt.bar(days, swiggy, bottom=zomato,
        color="orange", label="Swiggy")

plt.title("Daily Food Orders: Zomato vs Swiggy")
plt.xlabel("Days of the Week")
plt.ylabel("Number of Orders")
plt.legend()
plt.show()
```

```python
# Task 2.
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify = [500, 550, 530, 600, 650, 700, 680]
youtube = [800, 850, 820, 900, 950, 1000, 980]

plt.figure(figsize=(10, 5))

plt.fill_between(days, spotify, color="green",
                 alpha=0.4, label="Spotify")

plt.fill_between(days, youtube, color="red",
                 alpha=0.3, label="YouTube")

plt.plot(days, spotify, color="green", marker="o")
plt.plot(days, youtube, color="red", marker="o")

plt.title("Daily Active Users: Spotify vs YouTube")
plt.xlabel("Days")
plt.ylabel("Active Users")
plt.legend()
plt.show()
```

```python
# Task 3.
import matplotlib.pyplot as plt

days = list(range(1, 11))

avg_time = [32, 30, 35, 28, 34, 31, 36, 29, 33, 30]
std_dev = [3, 2, 4, 2, 3, 2, 5, 3, 4, 2]

plt.figure(figsize=(10, 5))

plt.errorbar(days,
    avg_time,
    yerr=std_dev,
    fmt="o-",
    color="blue",
    capsize=5,
    label="Average Delivery Time")

plt.title("Flipkart Average Delivery Time with Standard Deviation")
plt.xlabel("Day")
plt.ylabel("Delivery Time (Minutes)")
plt.xticks(days)
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Task 4.

import matplotlib.pyplot as plt
import pandas as pd
import matplotlib.dates as mdates

dates = pd.date_range(end=pd.Timestamp.today().normalize(),
    periods=14,
    freq="D")

balance = [2500, 2300, 2800, 2600, 3000, 2750, 3200,
          3100, 3500, 3300, 3700, 3400, 3900, 4200]

plt.figure(figsize=(10, 5))

plt.plot(dates, balance, marker="o", color="purple")

plt.title("Daily Paytm Wallet Balance")
plt.xlabel("Date")
plt.ylabel("Wallet Balance (₹)")

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Task 5.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify = [500, 550, 530, 600, 650, 700, 680]
youtube = [800, 850, 820, 900, 950, 1000, 980]

fig, ax = plt.subplots(figsize=(10, 5))

ax.fill_between(days, spotify,
    color="green", alpha=0.4,
    label="Spotify")

ax.fill_between(days, youtube,
    color="orange", alpha=0.3,
    label="YouTube")

ax.plot(days, spotify, color="green", marker="o")
ax.plot(days, youtube, color="orange", marker="o")

ax.set_title("Weekly Active Users Dashboard")
ax.set_xlabel("Day of the Week")
ax.set_ylabel("Daily Active Users")

ax.grid(True)
ax.legend()

plt.tight_layout()
plt.show()
```
