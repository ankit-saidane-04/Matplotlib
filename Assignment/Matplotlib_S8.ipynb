```python
# Task 1.
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
active_users = [1200, 1350, 1280, 1500, 1650, 1800, 1950]

plt.figure(figsize=(10, 4))

plt.plot(days, active_users, marker="o", color="blue")

plt.title("Daily Active Users - Spotify")
plt.xlabel("Days")
plt.ylabel("Active Users")

plt.grid(True)
plt.show()
```

```python
# Task 2.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato_orders = [120, 150, 140, 180, 200, 250, 230]
swiggy_orders = [100, 130, 160, 170, 190, 220, 260]

plt.plot(days, zomato_orders, linestyle="--",
         marker="o", label="Zomato")

plt.plot(days, swiggy_orders, linestyle=":",
         marker="s", label="Swiggy")

plt.title("Weekly Food Orders")
plt.xlabel("Days")
plt.ylabel("Number of Orders")

plt.legend()
plt.grid(True)
plt.show()
```

```python
# Task 3.

import matplotlib.pyplot as plt

days = list(range(1, 11))
sales = [12000, 15000, 13500, 18000, 17000,
         21000, 23000, 20000, 25000, 28000]

plt.style.use("ggplot")

plt.plot(days, sales, color="orange", marker="o")

plt.title("Flipkart Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales (₹)")

plt.show()
```

```python
# Task 4.
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
posts = [12, 18, 15, 22, 25, 30, 20]

instagram_colors = ["#833AB4",  # Purple
                   "#C13584",  # Magenta
                   "#E1306C",  # Pink
                   "#FD1D1D",  # Red
                   "#F56040",  # Orange-red
                   "#F77737",  # Orange
                   "#833AB4"   # Purple
                    ]

plt.figure(figsize=(9, 5))

plt.bar(days, posts, color=instagram_colors)

plt.title("Daily Instagram Posts")
plt.xlabel("Days")
plt.ylabel("Number of Posts")

plt.show()
```

```python
# Task 5.
import matplotlib.pyplot as plt

days = list(range(1, 11))
sales = [12000, 15000, 13500, 18000, 17000,
         21000, 23000, 20000, 25000, 28000]

plt.style.use("dark_background")

plt.figure(figsize=(10, 5))

plt.plot(days, sales, color="cyan",
         marker="D", linewidth=2)

plt.title("Flipkart Daily Sales - Dark Theme")
plt.xlabel("Day")
plt.ylabel("Sales (₹)")

plt.grid(True, alpha=0.3)
plt.show()
```
