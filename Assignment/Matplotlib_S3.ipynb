```python
import matplotlib.pyplot as plt
```

```python
# Task 1.

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [5000, 4500, 3000, 2000, 3500]

colors = ["red", "orange", "blue", "green", "purple"]

plt.bar(apps, orders, color=colors)

plt.title("Daily Orders by Food Delivery App")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")

plt.show()
```

```python
# Task 2.

influencers = ["Influencer A", "Influencer B", "Influencer C",
    "Influencer D", "Influencer E", "Influencer F"]

followers = [12, 8, 15, 6, 20, 10]

plt.barh(influencers, followers)

plt.title("Instagram Followers by Influencer")
plt.xlabel("Number of Followers")
plt.ylabel("Influencers")

plt.show()
```

```python
# Task 3.

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [5000, 4500, 3000, 2000, 3500]

colors = ["red", "orange", "blue", "green", "purple"]

plt.bar(apps, orders, color=colors, width=0.6)

for i in range(len(orders)):
    plt.text(i, orders[i] + 50, str(orders[i]), ha="center")

plt.title("Daily Orders by Food Delivery App")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")

plt.show()
```
