```python
# Task 1.

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6000, 7500, 5000, 8500, 10000, 9000]

plt.plot(days, steps, marker="o")

plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.title("Daily Steps Over 7 Days")

plt.grid(True)
plt.show()
```

```python
# Task 2.

posts = ["Post 1", "Post 2", "Post 3", "Post 4", "Post 5","Post 6", "Post 7", "Post 8", "Post 9", "Post 10"]

likes = [120, 180, 150, 250, 300, 220, 400, 350, 500, 450]

plt.plot(posts, likes, marker="o")

plt.xlabel("posts")
plt.ylabel("likes")
plt.title("instagram likes Analysis")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
# Task 3.
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_hours = [1.5, 2, 1, 2.5, 3, 4, 3.5]
youtube_hours = [1, 1.5, 2, 1, 2.5, 3, 2]

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(days, spotify_hours, marker="o")
ax[0].set_title("Daily Spotify Listening")
ax[0].set_xlabel("Days")
ax[0].set_ylabel("Listening Hours")

ax[1].plot(days, youtube_hours, marker="o")
ax[1].set_title("Daily YouTube Viewing")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Viewing Hours")

plt.tight_layout()
plt.show()
```

```python
# Task 4.

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_hours = [1.5, 2, 1, 2.5, 3, 4, 3.5]
youtube_hours = [1, 1.5, 2, 1, 2.5, 3, 2]

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(days, spotify_hours, marker="o")
ax[0].set_xlabel("Days")
ax[0].set_ylabel("Hours")
ax[0].set_title("Spotify Listening Hours")
ax[0].grid(True)

ax[1].plot(days, youtube_hours, marker="o")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Hours")
ax[1].set_title("YouTube Viewing Hours")
ax[1].grid(True)

fig.suptitle("Weekly Media Usage Dashboard")

plt.tight_layout()
plt.show()
```
