```python
import matplotlib.pyplot as plt
```

```python
# Task 1.

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [5000, 7000, 6500, 8000, 9000, 10000, 7500]
water = [2, 2.5, 2, 3, 2.5, 3.5, 3]

plt.figure(figsize=(8, 6))

# First subplot: 2 rows, 1 column, position 1
plt.subplot(2, 1, 1)
plt.plot(days, steps, color="green", marker="o")
plt.title("Daily Step Count")
plt.ylabel("Steps")
plt.grid(True)

# Second subplot: 2 rows, 1 column, position 2
plt.subplot(2, 1, 2)
plt.plot(days, water, color="blue", marker="o")
plt.title("Daily Water Intake")
plt.xlabel("Days")
plt.ylabel("Water (Litres)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Task 2.

import matplotlib.pyplot as plt

quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [25, 35, 20, 40]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# 1. Line chart
ax[0, 0].plot(quarters, posts, marker="o", color="blue")
ax[0, 0].set_title("Line Chart")
ax[0, 0].set_ylabel("Number of Posts")

# 2. Bar chart
ax[0, 1].bar(quarters, posts, color="orange")
ax[0, 1].set_title("Bar Chart")
ax[0, 1].set_ylabel("Number of Posts")

# 3. Scatter plot
ax[1, 0].scatter(quarters, posts, color="green", s=100)
ax[1, 0].set_title("Scatter Plot")
ax[1, 0].set_ylabel("Number of Posts")

# 4. Pie chart
ax[1, 1].pie(posts, labels=quarters, autopct="%1.1f%%")
ax[1, 1].set_title("Pie Chart")

fig.suptitle("Instagram Posts by Quarter")
plt.tight_layout()
plt.show()
```

```python
# Task 3.

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

zomato = [12, 15, 10, 18, 20, 16]
swiggy = [10, 12, 14, 15, 17, 19]
dominos = [5, 8, 7, 10, 12, 9]

fig, ax = plt.subplots(1, 3, figsize=(12, 4), sharey=True)

ax[0].plot(months, zomato, marker="o")
ax[0].set_title("Zomato Orders")
ax[0].set_ylabel("Number of Orders")

ax[1].plot(months, swiggy, marker="o", color="orange")
ax[1].set_title("Swiggy Orders")

ax[2].plot(months, dominos, marker="o", color="green")
ax[2].set_title("Domino's Orders")

for a in ax:
    a.set_xlabel("Months")
    a.grid(True)

fig.suptitle("Monthly Food Orders Comparison")
plt.tight_layout()
plt.show()
```

```python
# Task 4.
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

youtube = [8, 10, 12, 15, 14, 18]
instagram = [4, 5, 6, 7, 8, 9]
spotify = [2, 2.5, 3, 3.5, 4, 4.5]
whatsapp = [1, 1.5, 2, 2, 2.5, 3]

fig, ax = plt.subplots(2, 2, figsize=(10, 8), sharex=True)

# YouTube
ax[0, 0].plot(months, youtube, marker="o", color="red")
ax[0, 0].set_title("YouTube")
ax[0, 0].set_ylabel("Data Usage (GB)")

# Instagram
ax[0, 1].plot(months, instagram, marker="o", color="purple")
ax[0, 1].set_title("Instagram")
ax[0, 1].set_ylabel("Data Usage (GB)")

# Spotify
ax[1, 0].plot(months, spotify, marker="o", color="green")
ax[1, 0].set_title("Spotify")
ax[1, 0].set_xlabel("Months")
ax[1, 0].set_ylabel("Data Usage (GB)")

# WhatsApp
ax[1, 1].plot(months, whatsapp, marker="o", color="blue")
ax[1, 1].set_title("WhatsApp")
ax[1, 1].set_xlabel("Months")
ax[1, 1].set_ylabel("Data Usage (GB)")

for a in ax.flat:
    a.grid(True)

fig.suptitle("Monthly Mobile Data Usage by App")
plt.tight_layout()
plt.show()
```
