```python
# Task 1.
import matplotlib.pyplot as plt

restaurants = ["Restaurant A", "Restaurant B", "Restaurant C",
               "Restaurant D", "Restaurant E"]
orders = [120, 180, 90, 150, 200]

plt.bar(restaurants, orders, color="skyblue")

plt.title("Zomato Restaurant Orders")
plt.xlabel("Restaurants")
plt.ylabel("Number of Orders")
plt.xticks(rotation=20)

plt.savefig("zomato_orders.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Task 2.
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [5000, 6500, 7000, 6000, 8500, 10000, 9000]

plt.plot(days, steps, marker="o", color="green")

plt.title("Daily Step Count")
plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.grid(True)

plt.savefig("weekly_steps.svg", bbox_inches="tight")
plt.show()
```

```python
# Task 3.
import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
time_spent = [3, 4, 2, 1]

plt.pie(time_spent,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90)

plt.title("Daily Time Spent on Apps")

plt.savefig("app_usage.pdf", bbox_inches="tight")
plt.show()
```

```python
# Task 5.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]

plt.figure(figsize=(8, 5))
plt.plot(months, sales, marker="o")

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.savefig("sales_chart_150dpi.pdf",
            dpi=150, bbox_inches="tight")

plt.savefig("sales_chart_300dpi.pdf",
            dpi=300, bbox_inches="tight")

plt.show()
```
